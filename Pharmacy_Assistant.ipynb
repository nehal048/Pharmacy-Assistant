```python
import pandas as pd

# Dataset 1: Historical pharmacy sales
sales_url = "https://raw.githubusercontent.com/IndahAinka/retail_sales_of_a_pharmacy_in_Indonesia/main/transaction.csv"

# Dataset 2: Indian medicine catalogue
medicine_url = "https://raw.githubusercontent.com/junioralive/Indian-Medicine-Dataset/main/DATA/indian_medicine_data.csv"

sales_df = pd.read_csv(sales_url)
medicines_df = pd.read_csv(medicine_url)

print("SALES DATASET")
print("Shape:", sales_df.shape)
print("Columns:", sales_df.columns.tolist())
display(sales_df.head())

print("\nMEDICINE CATALOGUE")
print("Shape:", medicines_df.shape)
print("Columns:", medicines_df.columns.tolist())
display(medicines_df.head())
```

```python
# STEP 4: Inspect and validate both datasets

# Clean up column types
sales_df["TGL"] = pd.to_datetime(sales_df["TGL"], errors="coerce")
sales_df["QTY"] = pd.to_numeric(sales_df["QTY"], errors="coerce")

# 1. Sales data overview
print("========== SALES DATA CHECK ==========")
print("Date range:", sales_df["TGL"].min(), "to", sales_df["TGL"].max())
print("Unique medicines:", sales_df["KD_OBAT"].nunique())
print("Missing dates:", sales_df["TGL"].isna().sum())
print("Missing quantities:", sales_df["QTY"].isna().sum())
print("Zero or negative quantities:", (sales_df["QTY"] <= 0).sum())
print("Missing medicine IDs:", sales_df["KD_OBAT"].isna().sum())

print("\nMissing values by column:")
print(sales_df.isna().sum())

# 2. Check how frequently each medicine appears
medicine_counts = sales_df.groupby("KD_OBAT")["TGL"].count()

print("\n========== SALES HISTORY ==========")
print("Medicines with at least 30 transaction rows:",
      (medicine_counts >= 30).sum())
print("Medicines with at least 100 transaction rows:",
      (medicine_counts >= 100).sum())

print("\nMost frequently recorded medicines:")
display(medicine_counts.sort_values(ascending=False).head(10))

# 3. Check medicine catalogue quality
print("\n========== MEDICINE CATALOGUE CHECK ==========")
print("Duplicate medicine IDs:", medicines_df["id"].duplicated().sum())
print("Missing medicine names:", medicines_df["name"].isna().sum())
print("Missing composition 1:",
      medicines_df["short_composition1"].isna().sum())
print("Missing composition 2:",
      medicines_df["short_composition2"].isna().sum())

print("\nCatalogue sample:")
display(
    medicines_df[
        ["name", "manufacturer_name",
         "short_composition1", "short_composition2"]
    ].head(10)
)
```

```python
# STEP 5: Prepare daily sales data

# Keep valid dates and medicine IDs
sales_clean = sales_df.dropna(
    subset=["TGL", "KD_OBAT", "QTY"]
).copy()

# Inspect unusual quantities before filtering
print("Quantity distribution:")
display(sales_clean["QTY"].describe())

print("Non-positive quantity rows:")
display(
    sales_clean.loc[
        sales_clean["QTY"] <= 0,
        ["TGL", "KD_OBAT", "QTY", "NO_RESEP"]
    ].head(10)
)

# Aggregate transactions into daily quantities
daily_sales = (
    sales_clean.groupby(["TGL", "KD_OBAT"], as_index=False)
    .agg(daily_qty=("QTY", "sum"))
    .sort_values(["KD_OBAT", "TGL"])
)

print("\nDaily sales shape:", daily_sales.shape)
print("Date range:", daily_sales["TGL"].min(),
      "to", daily_sales["TGL"].max())
print("Medicines:", daily_sales["KD_OBAT"].nunique())

display(daily_sales.head(10))
```

```python

# STEP 6: Prepare data and train our first forecasting model

import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error

# 1. Exclude non-positive transactions for this initial demand experiment.
# These may represent returns or corrections and need separate investigation.
positive_sales = sales_df.loc[
    sales_df["QTY"] > 0,
    ["TGL", "KD_OBAT", "QTY"]
].copy()

# Select the 20 medicines with the most positive transaction records
top_medicines = (
    positive_sales["KD_OBAT"]
    .value_counts()
    .head(20)
    .index
    .tolist()
)

positive_sales = positive_sales[
    positive_sales["KD_OBAT"].isin(top_medicines)
]

# Aggregate positive quantities by medicine and day
daily_model = (
    positive_sales.groupby(["TGL", "KD_OBAT"], as_index=False)
    .agg(daily_qty=("QTY", "sum"))
)

# 2. Create a complete daily timeline for each selected medicine.
# Missing dates are provisionally treated as zero recorded sales.
all_dates = pd.date_range(
    sales_df["TGL"].min(),
    sales_df["TGL"].max(),
    freq="D"
)

complete_parts = []

for medicine in top_medicines:
    temp = (
        daily_model[daily_model["KD_OBAT"] == medicine]
        .set_index("TGL")["daily_qty"]
        .reindex(all_dates, fill_value=0)
        .rename("daily_qty")
        .to_frame()
    )

    temp["KD_OBAT"] = medicine
    temp.index.name = "TGL"
    complete_parts.append(temp.reset_index())

model_data = pd.concat(complete_parts, ignore_index=True)
model_data = model_data.sort_values(["KD_OBAT", "TGL"])

# 3. Create features using past information only
model_data["day_of_week"] = model_data["TGL"].dt.dayofweek
model_data["month"] = model_data["TGL"].dt.month

for lag in [1, 7, 14]:
    model_data[f"lag_{lag}"] = (
        model_data.groupby("KD_OBAT")["daily_qty"].shift(lag)
    )

model_data["rolling_mean_7"] = (
    model_data.groupby("KD_OBAT")["daily_qty"]
    .transform(lambda s: s.shift(1).rolling(7).mean())
)

model_data["rolling_mean_14"] = (
    model_data.groupby("KD_OBAT")["daily_qty"]
    .transform(lambda s: s.shift(1).rolling(14).mean())
)

feature_cols = [
    "day_of_week", "month",
    "lag_1", "lag_7", "lag_14",
    "rolling_mean_7", "rolling_mean_14"
]

model_data = model_data.dropna(subset=feature_cols).copy()

# 4. Chronological train/test split: final 60 days for testing
cutoff_date = model_data["TGL"].max() - pd.Timedelta(days=59)

train = model_data[model_data["TGL"] < cutoff_date]
test = model_data[model_data["TGL"] >= cutoff_date]

X_train = train[feature_cols]
y_train = train["daily_qty"]
X_test = test[feature_cols]
y_test = test["daily_qty"]

# 5. Train the model
model = RandomForestRegressor(
    n_estimators=100,
    max_depth=12,
    min_samples_leaf=3,
    random_state=42,
    n_jobs=-1
)

model.fit(X_train, y_train)
predictions = np.maximum(model.predict(X_test), 0)

# 6. Evaluate against a simple weekly baseline
baseline_predictions = test["lag_7"].clip(lower=0)

mae = mean_absolute_error(y_test, predictions)
rmse = np.sqrt(mean_squared_error(y_test, predictions))

baseline_mae = mean_absolute_error(y_test, baseline_predictions)

print("========== MODEL RESULTS ==========")
print("Medicines:", len(top_medicines))
print("Training rows:", len(train))
print("Testing rows:", len(test))
print("Test period:", test["TGL"].min().date(),
      "to", test["TGL"].max().date())
print(f"Random Forest MAE: {mae:.3f}")
print(f"Random Forest RMSE: {rmse:.3f}")
print(f"Weekly baseline MAE: {baseline_mae:.3f}")

# 7. Show some actual vs predicted sales
results = test[["TGL", "KD_OBAT", "daily_qty"]].copy()
results["predicted_qty"] = predictions
results["weekly_baseline"] = baseline_predictions.to_numpy()

display(results.head(15))
```

```python

# STEP 7: Generate 7-day demand forecasts

forecast_horizon = 7
forecast_rows = []

for medicine in top_medicines:
    history = (
        model_data.loc[
            model_data["KD_OBAT"] == medicine,
            ["TGL", "daily_qty"]
        ]
        .sort_values("TGL")
        .set_index("TGL")["daily_qty"]
        .copy()
    )

    # Include all dates through the end of the historical dataset
    history = history.reindex(
        pd.date_range(
            sales_df["TGL"].min(),
            sales_df["TGL"].max(),
            freq="D"
        ),
        fill_value=0
    )

    history.index.name = "TGL"

    for step in range(1, forecast_horizon + 1):
        next_date = history.index.max() + pd.Timedelta(days=1)

        features = pd.DataFrame([{
            "day_of_week": next_date.dayofweek,
            "month": next_date.month,
            "lag_1": history.iloc[-1],
            "lag_7": history.iloc[-7],
            "lag_14": history.iloc[-14],
            "rolling_mean_7": history.iloc[-7:].mean(),
            "rolling_mean_14": history.iloc[-14:].mean()
        }])[feature_cols]

        predicted_qty = max(0, float(model.predict(features)[0]))

        forecast_rows.append({
            "date": next_date,
            "medicine_id": medicine,
            "predicted_demand": round(predicted_qty, 2)
        })

        # Feed the prediction back into history for the next day
        history.loc[next_date] = predicted_qty

forecast_df = pd.DataFrame(forecast_rows)

print("7-day forecast generated!")
display(forecast_df.head(20))

print("\nTotal forecast rows:", len(forecast_df))
print("Forecast dates:",
      forecast_df["date"].min().date(),
      "to",
      forecast_df["date"].max().date())
```

```python

# STEP 8: Inventory and low-stock alerts
import numpy as np
import pandas as pd

# Reproducible DEMO data only — not actual pharmacy inventory
rng = np.random.default_rng(42)

inventory_df = pd.DataFrame({
    "medicine_id": top_medicines,
    "current_stock": rng.integers(50, 1200, size=len(top_medicines)),
    "minimum_stock": rng.integers(50, 200, size=len(top_medicines))
})

# Total predicted demand across the next 7 days
weekly_demand = (
    forecast_df.groupby("medicine_id")["predicted_demand"]
    .sum()
    .reset_index(name="predicted_7_day_demand")
)

inventory_df = inventory_df.merge(
    weekly_demand, on="medicine_id", how="left"
)

# Flag low stock and forecasted shortages separately
inventory_df["low_stock"] = (
    inventory_df["current_stock"] <= inventory_df["minimum_stock"]
)

inventory_df["shortage_risk"] = (
    inventory_df["current_stock"]
    < inventory_df["predicted_7_day_demand"]
)

inventory_df["stock_after_forecast"] = (
    inventory_df["current_stock"]
    - inventory_df["predicted_7_day_demand"]
)

inventory_df["status"] = np.select(
    [
        inventory_df["shortage_risk"],
        inventory_df["low_stock"]
    ],
    [
        "Forecast shortage risk",
        "Below minimum stock"
    ],
    default="Stock currently sufficient"
)

inventory_df = inventory_df.sort_values(
    ["shortage_risk", "low_stock"],
    ascending=False
).reset_index(drop=True)

print("NOTE: All inventory quantities below are SAMPLE data.")
print("Medicines at shortage risk:",
      inventory_df["shortage_risk"].sum())
print("Medicines below minimum stock:",
      inventory_df["low_stock"].sum())

display(inventory_df)
```

```python

# STEP 9: Find medicines with matching listed compositions

import re
import pandas as pd

catalogue = medicines_df.copy()

# Exclude discontinued products from suggestions
catalogue = catalogue[
    catalogue["Is_discontinued"].astype(str).str.lower().isin(["false", "0"])
].copy()

def normalize_ingredient(value):
    """Normalize whitespace and capitalization for ingredient matching."""
    if pd.isna(value):
        return ""
    return re.sub(r"\s+", " ", str(value)).strip().lower()

def composition_key(row):
    """Use both listed composition fields, retaining strength."""
    ingredients = [
        normalize_ingredient(row["short_composition1"]),
        normalize_ingredient(row["short_composition2"])
    ]
    ingredients = sorted(x for x in ingredients if x)
    return tuple(ingredients)

catalogue["composition_key"] = catalogue.apply(composition_key, axis=1)

def find_alternatives(medicine_name, limit=10):
    selected = catalogue[
        catalogue["name"].str.casefold() == medicine_name.casefold()
    ]

    if selected.empty:
        print("Medicine not found in the catalogue.")
        return pd.DataFrame()

    key = selected.iloc[0]["composition_key"]

    if not key:
        print("Composition information is unavailable.")
        return pd.DataFrame()

    alternatives = catalogue[
        (catalogue["composition_key"] == key)
        & (catalogue["name"].str.casefold() != medicine_name.casefold())
    ].copy()

    columns = [
        "name", "manufacturer_name", "price(₹)",
        "pack_size_label", "short_composition1", "short_composition2"
    ]

    if alternatives.empty:
        print("No exact listed-composition matches found.")
        return pd.DataFrame(columns=columns)

    return alternatives[columns].head(limit)

# Example: test with a medicine from our catalogue
example_medicine = "Augmentin 625 Duo Tablet"

print("Looking for matching listed compositions for:", example_medicine)
display(find_alternatives(example_medicine))
```

```python
# STEP 10: Prepare catalogue-based inventory

catalogue_inventory = catalogue.reset_index(drop=True).copy()

# Keep medicines with listed composition information
catalogue_inventory = catalogue_inventory[
    catalogue_inventory["composition_key"].apply(bool)
].copy()

# Give each catalogue medicine a stable identifier
catalogue_inventory["catalogue_id"] = (
    catalogue_inventory["id"].astype(str)
)

# Create SAMPLE stock values for the project demo
rng = np.random.default_rng(42)

catalogue_inventory["current_stock"] = rng.integers(
    5, 101, size=len(catalogue_inventory)
)

catalogue_inventory["minimum_stock"] = rng.integers(
    10, 31, size=len(catalogue_inventory)
)

catalogue_inventory["stock_status"] = np.where(
    catalogue_inventory["current_stock"]
    <= catalogue_inventory["minimum_stock"],
    "Low Stock",
    "In Stock"
)

print("Catalogue medicines:", len(catalogue_inventory))
print("\nStock status counts:")
print(catalogue_inventory["stock_status"].value_counts())

display(
    catalogue_inventory[
        [
            "catalogue_id",
            "name",
            "current_stock",
            "minimum_stock",
            "stock_status"
        ]
    ].head(10)
)
```

```python
# STEP 11: Find alternatives for a low-stock medicine

def show_low_stock_alternatives(catalogue_id, limit=10):
    catalogue_id = str(catalogue_id)

    selected = catalogue_inventory[
        catalogue_inventory["catalogue_id"] == catalogue_id
    ]

    if selected.empty:
        print("Medicine ID not found.")
        return

    medicine = selected.iloc[0]

    print("Selected medicine:", medicine["name"])
    print("Current stock:", medicine["current_stock"])
    print("Minimum stock:", medicine["minimum_stock"])
    print("Status:", medicine["stock_status"])

    if medicine["stock_status"] != "Low Stock":
        print("\nThis medicine is not currently flagged as low stock.")
        return

    key = medicine["composition_key"]

    alternatives = catalogue_inventory[
        (catalogue_inventory["composition_key"] == key)
        & (catalogue_inventory["catalogue_id"] != catalogue_id)
    ].copy()

    columns = [
        "catalogue_id",
        "name",
        "manufacturer_name",
        "price(₹)",
        "pack_size_label",
        "current_stock",
        "stock_status"
    ]

    print("\nPotential matching catalogue entries:")
    if alternatives.empty:
        print("No matching entries found.")
        return

    display(alternatives[columns].head(limit))


# Demo: Augmentin 625 Duo Tablet has catalogue ID 1
show_low_stock_alternatives("1")
```

```python

import gradio as gr
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# DASHBOARD FUNCTION
# -----------------------------
def dashboard():
    total = len(catalogue_inventory)

    low = catalogue_inventory[
        catalogue_inventory["stock_status"] == "Low Stock"
    ]
    in_stock = catalogue_inventory[
        catalogue_inventory["stock_status"] == "In Stock"
    ]

    summary = pd.DataFrame({
        "Metric": ["Total medicines", "Low stock", "In stock"],
        "Count": [total, len(low), len(in_stock)]
    })

    fig, ax = plt.subplots(figsize=(7, 4))
    ax.bar(
        ["In Stock", "Low Stock"],
        [len(in_stock), len(low)]
    )
    ax.set_title("Inventory Stock Status")
    ax.set_ylabel("Number of catalogue entries")
    fig.tight_layout()

    attention = low[
        ["catalogue_id", "name", "current_stock",
         "minimum_stock", "stock_status"]
    ].sort_values("current_stock").head(15)

    return summary, fig, attention


# -----------------------------
# MEDICINE SEARCH
# -----------------------------
def search_medicines(query):
    columns = [
        "catalogue_id", "name", "price(₹)",
        "current_stock", "minimum_stock", "stock_status"
    ]

    if not query or not query.strip():
        return catalogue_inventory[columns].head(50)

    mask = catalogue_inventory["name"].astype(str).str.contains(
        query.strip(), case=False, na=False
    )

    return catalogue_inventory.loc[mask, columns].head(50)


# -----------------------------
# COMPOSITION MATCHING
# -----------------------------
def find_matches(medicine_name):
    columns = [
        "catalogue_id", "name", "manufacturer_name",
        "short_composition1", "short_composition2",
        "current_stock", "stock_status"
    ]

    if not medicine_name or not medicine_name.strip():
        return pd.DataFrame({
            "Message": ["Enter a medicine name first."]
        })

    selected = catalogue_inventory[
        catalogue_inventory["name"].astype(str).str.contains(
            medicine_name.strip(), case=False, na=False
        )
    ]

    if selected.empty:
        return pd.DataFrame({
            "Message": ["No medicine found. Try a shorter name."]
        })

    row = selected.iloc[0]
    key = row["composition_key"]

    if not key:
        return pd.DataFrame({
            "Message": ["Composition information is unavailable."]
        })

    matches = catalogue_inventory[
        catalogue_inventory["composition_key"].apply(
            lambda x: bool(x) and x == key
        )
    ].copy()

    matches = matches[
        matches["catalogue_id"] != row["catalogue_id"]
    ]

    if matches.empty:
        return pd.DataFrame({
            "Message": ["No other entries share this composition."]
        })

    return (
        matches[columns]
        .drop_duplicates()
        .head(20)
        .reset_index(drop=True)
    )


# -----------------------------
# HEALTHCARE THEME
# -----------------------------
medical_theme = gr.themes.Soft(
    primary_hue="teal",
    secondary_hue="emerald",
    neutral_hue="gray",
    font=["Inter", "Arial", "sans-serif"],
    font_mono=["Consolas", "monospace"],
    radius_size="lg",
    spacing_size="lg"
)



custom_css = """
/* Overall page */
.gradio-container {
    background: #E4F1EA !important;
    max-width: 100% !important;
    width: 100% !important;
    margin: auto !important;
    padding: 12px 20px !important;
    box-sizing: border-box !important;
}

/* Main Pharmacy Assistant header card */
.home-header {
    background: #F7FBF8 !important;
    border: 1px solid #C9E0D4 !important;
    border-radius: 20px !important;
    padding: 28px !important;
    margin-bottom: 8px !important;
}

/* Header text: dark green and easy to read */
.home-header h1 {
    color: #205746 !important;
    font-size: 38px !important;
    font-weight: 800 !important;
}

.home-header h2,
.home-header h3 {
    color: #326C57 !important;
    font-size: 23px !important;
    font-weight: 700 !important;
}

.home-header p,
.home-header strong {
    color: #365D4F !important;
    font-size: 17px !important;
    line-height: 1.6 !important;
}

/* Navigation instructions: dark green, never pale */
.gradio-container h2,
.gradio-container h3 {
    color: #205746 !important;
}

.gradio-container p {
    color: #365D4F !important;
}

/* Cards: GREEN BACKGROUND, DARK TEXT */
.nav-card button {
    background: #C7E0D2 !important;
    color: #214D40 !important;
    border: 1px solid #AFCFBE !important;
    border-radius: 18px !important;
    min-height: 180px !important;
    height: 180px !important;
    width: 100% !important;
    white-space: pre-wrap !important;
    text-align: center !important;
    font-size: 19px !important;
    font-weight: 650 !important;
    box-shadow: 0 4px 12px rgba(31, 91, 72, 0.08);
    transition: all 0.2s ease;
}

.nav-card button:hover {
    background: #B8D8C6 !important;
    color: #214D40 !important;
    border-color: #83B69A !important;
    transform: translateY(-2px);
}

/* Reduce unnecessary gaps */
.nav-card {
    min-width: 0 !important;
}

footer {
    display: none !important;
}

/* Mobile layout */
@media (max-width: 700px) {
    .gradio-container {
        padding: 10px !important;
    }

    .home-header h1 {
        font-size: 29px !important;
    }

    .nav-card button {
        min-height: 145px !important;
        height: 145px !important;
        font-size: 16px !important;
    }
}
"""




# -----------------------------
# PHARMACY ASSISTANT APP
# -----------------------------
with gr.Blocks(
    title="Pharmacy Assistant",
    theme=medical_theme,
    css=custom_css
) as app:



    pages = []

    # HOME SCREEN
    with gr.Column(visible=True) as home_page:
        with gr.Column(elem_classes=["home-header"]):
            gr.Markdown("""
            # 💊 Pharmacy Assistant
            ### Your smart pharmacy workspace

            Manage medicine inventory, review stock alerts,
            explore composition matches, and view demand forecasts.
            """)

        gr.Markdown("""
## What would you like to do today?
<p style="color:#365D4F; font-size:17px;">
Choose a section below to get started.
</p>
""")
        with gr.Row(equal_height=True):
            with gr.Column(elem_classes=["nav-card"]):
                overview_card = gr.Button(
                    "🏥\n\nInventory Overview\n\nReview stock levels and alerts",
                    size="xl"
                )

            with gr.Column(elem_classes=["nav-card"]):
                inventory_card = gr.Button(
                    "📦\n\nMedicine Inventory\n\nSearch medicines and check stock",
                    size="xl"
                )

        with gr.Row(equal_height=True):
            with gr.Column(elem_classes=["nav-card"]):
                matches_card = gr.Button(
                    "💊\n\nComposition Matches\n\nFind matching listed ingredients and strengths",
                    size="xl"
                )

            with gr.Column(elem_classes=["nav-card"]):
                forecast_card = gr.Button(
                    "📈\n\nDemand Forecast\n\nExplore model predictions",
                    size="xl"
                )

    # OVERVIEW PAGE
    with gr.Column(visible=False) as overview_page:
        back_overview = gr.Button("← Home")
        gr.Markdown("## 🏥 Inventory Overview")

        refresh = gr.Button("Refresh Overview", variant="primary")
        summary = gr.Dataframe(
            label="Inventory Summary",
            interactive=False
        )
        chart = gr.Plot(label="Stock Distribution")
        low_stock = gr.Dataframe(
            label="Medicines Requiring Attention",
            interactive=False
        )

        refresh.click(
            fn=dashboard,
            outputs=[summary, chart, low_stock]
        )

        app.load(
            fn=dashboard,
            outputs=[summary, chart, low_stock]
        )

    # MEDICINE INVENTORY PAGE
    with gr.Column(visible=False) as inventory_page:
        back_inventory = gr.Button("← Home")
        gr.Markdown("## 📦 Medicine Inventory")
        gr.Markdown(
            "Search the medicine catalogue and review simulated stock levels."
        )

        with gr.Row():
            search_box = gr.Textbox(
                label="Medicine Name",
                placeholder="e.g. A-Bec Tablet"
            )
            search_button = gr.Button(
                "Search Medicines",
                variant="primary"
            )

        inventory_results = gr.Dataframe(
            label="Search Results",
            interactive=False
        )

        search_button.click(
            fn=search_medicines,
            inputs=search_box,
            outputs=inventory_results
        )

        search_box.submit(
            fn=search_medicines,
            inputs=search_box,
            outputs=inventory_results
        )

        gr.Markdown(
            "*Inventory quantities are simulated for this project.*"
        )

    # COMPOSITION MATCHES PAGE
    with gr.Column(visible=False) as matches_page:
        back_matches = gr.Button("← Home")
        gr.Markdown("## 💊 Medicine Composition Matches")
        gr.Markdown(
            "Find catalogue entries sharing the same recorded ingredients "
            "and strengths."
        )

        medicine_input = gr.Textbox(
            label="Medicine Name",
            placeholder="e.g. A-Bec Tablet"
        )

        alternatives_button = gr.Button(
            "Find Composition Matches",
            variant="primary"
        )

        alternatives_output = gr.Dataframe(
            label="Potential Composition Matches",
            interactive=False
        )

        alternatives_button.click(
            fn=find_matches,
            inputs=medicine_input,
            outputs=alternatives_output
        )

        gr.Markdown(
            "*Matching composition does not guarantee clinical "
            "interchangeability. A pharmacist must verify suitability.*"
        )

    # DEMAND FORECAST PAGE
    with gr.Column(visible=False) as forecast_page:
        back_forecast = gr.Button("← Home")
        gr.Markdown("## 📈 Demand Forecast")
        gr.Markdown(
            "Review the output from your existing forecasting model."
        )

        if "forecast_df" in globals():
            gr.Dataframe(
                value=forecast_df.head(50),
                label="Predicted Demand",
                interactive=False
            )
        else:
            gr.Markdown(
                "Forecast data is unavailable. Run the forecasting cells first."
            )

        gr.Markdown(
            "*These experimental predictions use historical 2015 sales "
            "data and cover January 1–7, 2016. They do not represent "
            "current pharmacy demand.*"
        )

    # PAGE NAVIGATION
    pages = [
        home_page,
        overview_page,
        inventory_page,
        matches_page,
        forecast_page
    ]

    def navigate(destination):
        return tuple(
            gr.update(visible=(i == destination))
            for i in range(len(pages))
        )

    overview_card.click(
        lambda: navigate(1),
        outputs=pages
    )
    inventory_card.click(
        lambda: navigate(2),
        outputs=pages
    )
    matches_card.click(
        lambda: navigate(3),
        outputs=pages
    )
    forecast_card.click(
        lambda: navigate(4),
        outputs=pages
    )

    back_overview.click(lambda: navigate(0), outputs=pages)
    back_inventory.click(lambda: navigate(0), outputs=pages)
    back_matches.click(lambda: navigate(0), outputs=pages)
    back_forecast.click(lambda: navigate(0), outputs=pages)

    gr.Markdown("---\n*Pharmacy Assistant · Academic ML Project*")

app.launch(share=False)
```
